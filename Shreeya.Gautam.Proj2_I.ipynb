Dataset Name: **Global Environmental Trends 2020-2024**

The dataset was **Kaggle** (https://www.kaggle.com/datasets/adilshamim8/temperature/data)

This dataset offers a detailed breakdown of critical environmental indicators obtained from several nations over 24 years (2000–2024). It is intended to assist with analyses that investigate how human activity, environmental changes, and climate variables interact.


# Dataset Overview and Structure

```python
import pandas as pd
#Load dataset 
df = pd.read_csv('temperature.csv')
```

# 1. To find shape and structure 

```python
print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nDataFrame Info:")
df.info()
```

# 2. Identify variable types

```python
print("\nData Types:")
print(df.dtypes)

df.dtypes.value_counts()
```

# 3. Check for Missing Values

```python
missing = df.isnull().sum()
total_missing = missing.sum()

print(f"\nTotal Missing Values: {total_missing}")
print("\nMissing Values by Column:")
print(missing[missing > 0])

#Plan to handle missing values
# Strategy: Drop rows with missing data (for now, may change later in analysis)
df = df.dropna()
print("\nNew shape after dropping rows with missing values:", df.shape)
```

# **SECTION II: DATA CLEANING & INDEXING**
# 4. Check and remove duplicates

```python
duplicates = df.duplicated().sum()
print("Number of duplicate rows:", duplicates)

if duplicates > 0:
    df = df.drop_duplicates()
    print("Duplicates removed.")
```

# 5. Two filtering examples

```python
# Example 1: Countries with CO2 emissions > 10 tons per capita
high_emission = df[df['CO2_Emissions_tons_per_capita'] > 10]
print("Countries with high CO2 emissions per capita:")
display(high_emission[['Country', 'Year', 'CO2_Emissions_tons_per_capita']].head())

# Example 2: Records with Renewable Energy usage above 50%
renewable_high = df[df['Renewable_Energy_pct'] > 50]
print("Countries with >50% renewable energy usage:")
display(renewable_high[['Country', 'Year', 'Renewable_Energy_pct']].head())
```

# 6. Summary stats for numerical columns

```python
desc = df.describe()
display(desc)

# Insight example:
print("\nInsight: The maximum number of extreme weather events recorded in a single year-country pair is", df['Extreme_Weather_Events'].max())
```

# 7. Value counts for categorical columns

```python
# Country frequency
print("Top 5 Countries by record count:")
print(df['Country'].value_counts().head())

# Year frequency
print("\nYears of observation:")
print(df['Year'].value_counts().sort_index())
```

# 8. Groupby analysis

```python
# Average CO2 emissions by country
co2_avg_by_country = df.groupby('Country')['CO2_Emissions_tons_per_capita'].mean().sort_values(ascending=False)
print("Top 5 countries by average CO2 emissions:")
display(co2_avg_by_country.head())

# Mean renewable energy usage by year
renewable_by_year = df.groupby('Year')['Renewable_Energy_pct'].mean()
print("\nAverage Renewable Energy Usage by Year:")
display(renewable_by_year.head())
```

# For cleaned Dataset

```python
# Preview cleaned dataset
print("Cleaned Dataset Shape:", df.shape)
display(df.head())

# Confirm that no missing values are remaining
print("\nRemaining Missing Values by Column:")
print(df.isnull().sum())

df.to_csv("cleaned_environmental_trends_2000_2024.csv", index=False)
print("Cleaned dataset saved as 'cleaned_environmental_trends_2000_2024.csv'")
```
